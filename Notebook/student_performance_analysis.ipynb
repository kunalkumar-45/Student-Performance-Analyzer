## 8. Correlation Analysis

Correlation analysis was used to examine the relationships between Attendance, Study Hours, and Average Score.

The results show very strong positive correlations between these variables.
Attendance has the strongest correlation with Average Score (0.996),
while Study Hours also show a very strong positive correlation with Average Score (0.983).

Correlation indicates a relationship between variables but does not prove causation.
